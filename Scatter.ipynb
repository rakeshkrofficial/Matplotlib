```python
import matplotlib.pyplot as plt
```

```python
studies_hours=[1,2,3,4,5,6,7,8]
exam_scores =[50,55,60,65,70,75,80,85]

plt.scatter(studies_hours, exam_scores, color='red',marker='*',label="Student Data")

plt.xlabel("Hours Studies")
plt.ylabel("Exam Scores")

plt.title("Rekationship Between Study Time and Exam Score")
plt.legend()

plt.grid(color='gray',linestyle=':',linewidth=1)
```

```python
plt.scatter([1,2,3],[40,50,65],color='red',marker='o',label='Class A')
plt.scatter([1,2,3],[35,43,50],color='blue',marker='o',label='class B')

plt.title("Hours Studies")
plt.xlabel("Hours")
plt.ylabel("Marks")

plt.grid(color='gray',linewidth=1,linestyle=':')
plt.legend()

plt.show()
```
