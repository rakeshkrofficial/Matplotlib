```python
import matplotlib.pyplot as plt
```

```python
x=["Mon","Tues","Wedy","Thur","Fri","Sat","Sun"]
y=[20,15,25,18,30,28,40]
plt.plot(x,y)

plt.title("Bakery Sales This Week!")

plt.xlabel("Day Of The Week")
plt.ylabel("Sales Per Day")

plt.show()
```

```python
a=[1,2,3,4,5]
b=[10,20,15,25,23]
plt.plot(a,b)

plt.title("Random chart for practice")
plt.xlabel("Days of the week")
plt.ylabel("Price of expensive pen")
```

```python
months=[1,2,3,4,5]
sales=[1500,2000,1800,2500,2300]

plt.plot(months,sales,color='orange',linestyle='--',linewidth=2,marker='o',label="Sales record in 2025")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales Data Report")
plt.legend()

plt.show()
```

```python
days=["Mon","Tues","Wed","Thur","Fri","sat","Sun"]
sales=[1000,900,1300,1200,1500,1400,2000]

plt.plot(days,sales,color='blue',linestyle='--',linewidth=2,marker='o',label="Day sales")
plt.legend(loc='lower right',fontsize=15)
plt.title("Per Day Sales Report")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid(color='gray',linestyle=':',linewidth=1)
```

```python
months=[1,2,3,4,5,6,7]
sales=[1200,1700,1500,2000,1800,2500,2300]

plt.plot(months,sales,linestyle='--',linewidth=2,marker='o',color='green',label=" Monthly Sales")
plt.legend(loc='upper left',fontsize=12)
plt.grid(linestyle=':',linewidth=1,color='gray')

plt.xlim(1,7)
plt.ylim(0,2600)

plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Report in Months!")

plt.xticks([1,2,3,4,5,6,7],["M1","M2","M3","M4","M5","M6","M7"])

plt.show()
```
