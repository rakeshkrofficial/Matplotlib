```python
import matplotlib.pyplot as plt
```

```python
product=["A","B","C","D","E"]
sales=[2000,1500,1800,2500,1200]

plt.bar(product,sales,color="skyblue",label="sales 2026") # barh is used to horigendal

plt.legend()

 # plt.grid(color='gray',linestyle=':',linewidth=1)   it is optional 
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product sales Record in 2026")
```

Pie Chart

```python
regions=["North","South","West","East"]
revenue=[2800,1500,2000,2500]

plt.pie(revenue,labels=regions, autopct='%1.1f%%',colors=['gold','skyblue','lightgreen','coral'])
plt.title("Revenue Contribution By Region")
plt.show()
```

```python
scores=[45,55,56,70,68,77,85,60,75,88,63,84,78,89,90,99,50,86,62,59]

plt.hist(scores, color='skyblue',edgecolor='black',bins=5)

plt.title("Scores Distribution of Students")
plt.xlabel("Scores Range")
plt.ylabel("Number of students")
plt.show()
```
