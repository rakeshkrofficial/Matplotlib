```python
import matplotlib.pyplot as plt
```

```python
marks = [87,89,79,72,93,65]
subject = ['Science','SST','Math','English','Hindi','Sanskrit']

plt.plot(subject,marks,linestyle=':',marker='o',linewidth=2,color='g',label="10th Marks")

plt.title("Graph of 10th marks")
plt.xlabel("Subject")
plt.ylabel("Marks")

plt.xlim(-0.5,6)
plt.ylim(49,100)

plt.legend(loc='lower left',fontsize=9)

plt.yticks([87,89,79,72,93,65,98])

plt.grid(linestyle=':',linewidth='2',color='black')
plt.show()
```
