```python
x=[1,2,3,4,5]
y=[1000,2000,1500,3000,2500]

plt.bar(x,y, color='green',label='Product sales 2025')

plt.title('2025 sales product in Months')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.legend()

plt.savefig('Figures_photo//bar_chart.png', dpi=300,bbox_inches='tight')
plt.show()
```
