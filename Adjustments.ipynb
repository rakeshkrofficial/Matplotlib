```python
import matplotlib.pyplot as plt
```

SUB Plot

```python
# plt.subplot(nrows, ncols, index)
```

```python
x=[1,2,3,4,5]
y=[10,20,15,25,20]

plt.subplot(1,2,1) # 1 row , 2 column , 1 suplot
plt.plot(x,y)
plt.title("Line Chart")

plt.subplot(1,2,2)# 1 row , 2 column , 2 subplot
plt.bar(x,y)
plt.title("Bar chart")
# plt.tight_layout()
```

```python
a=[1,2,3,4,5]
b=[25,40,35,50,45]

plt.subplot(1,2,1)
plt.bar(a,b)
plt.title("Line Chart")

plt.subplot(1,2,2)
plt.plot(a,b)
plt.title("Bar Chart")

plt.show()
```

```python
x=[1,2,3,4]
y=[25,20,30,40]

fig,ax=plt.subplots(1,2,figsize=[10,5])

ax[0].plot(x,y,color='skyblue')
ax[0].set_title("Line Plot")


ax[1].bar(x,y,color='green')
ax[1].set_title("Bar Chart")

fig.suptitle("Comparision of Line Chart and Bar Chart")

plt.tight_layout()
plt.show()
```
